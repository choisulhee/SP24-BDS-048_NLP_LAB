## Task 08: Functions and Lambda for NLP Pipelines
**Topic:** Functions, Lambda Reusable Text Processing

### (a)
Write a regular function called `clean(text)` that takes a string, strips whitespace, converts it to lowercase, and returns the result. Then write an equivalent lambda function called `clean_lambda` that performs the same two operations in one line. Test both on the string "  HELLO Natural Language Processing  " and print their outputs.

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample = "  HELLO Natural Language Processing  "

print(clean(sample))
print(clean_lambda(sample))
```

### (b)
Write a function called `tokenize(text)` that takes a cleaned string and returns a list of individual word tokens using `split()`. Then write another function called `remove_short(tokens, n)` that takes the token list and an integer `n`, and returns only tokens whose length is greater than `n`. Chain both functions on the sentence "I love working with NLP tools in Python daily" with n = 3 and print the result.

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"

token_list = tokenize(sentence)
result = remove_short(token_list, 3)
print(result)
```

### (c)
Write a function called `word_lengths(tokens)` that takes a list of tokens and returns a dictionary where each key is a word and each value is the length of that word. Call the function on `tokens = ["language", "model", "ai", "processing", "deep"]` and print the resulting dictionary.

```python
def word_lengths(tokens):
    lengths = {}
    for t in tokens:
        lengths[t] = len(t)
    return lengths

tokens = ["language", "model", "ai", "processing", "deep"]
result = word_lengths(tokens)
print(result)
```

### (d)
Combine the functions from parts (b) and (c) into a full pipeline function called `nlp_pipeline(text, min_len)` that: cleans the text, tokenises it, removes short words, and returns a length dictionary. Call this pipeline on two different sentences and print the outputs.

```python
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

sentence1 = "  Natural Language Processing is a fascinating field of AI  "
sentence2 = "Python makes it easy to build NLP pipelines fast"

print(nlp_pipeline(sentence1, 3))
print(nlp_pipeline(sentence2, 3))
```

## Task 09: map(), filter(), sorted() and Counter
**Topic:** map(), filter(), sorted(), Lambda, collections.Counter

### (a)
Given `words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]`. Use `map()` with a lambda to: (i) convert every word to lowercase - print the result as a list, and (ii) compute the character length of each word - print the result as a list.

```python
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lower_words = list(map(lambda w: w.lower(), words))
print(lower_words)

word_lens = list(map(lambda w: len(w), words))
print(word_lens)
```

### (b)
Using the same `words` list, apply `filter()` with a lambda to create two filtered lists: (i) `words_long` - only words with more than 3 characters, and (ii) `words_with_t` - only words containing the letter "t" (case-insensitive). Print both lists.

```python
words_long = list(filter(lambda w: len(w) > 3, words))
print(words_long)

words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print(words_with_t)
```

### (c)
Sort the `words` list using `sorted()` in three different ways and print each result with a label: (i) alphabetically (default order), (ii) by word length ascending using `key=len`, (iii) by word length descending using `reverse=True`.

```python
alphabetical = sorted(words)
print("Alphabetical:", alphabetical)

by_length_asc = sorted(words, key=len)
print("By length ascending:", by_length_asc)

by_length_desc = sorted(words, key=len, reverse=True)
print("By length descending:", by_length_desc)
```

### (d)
Import `Counter` from `collections`. Given the sentence "nlp nlp python ai nlp python text ai nlp python text", split it into words and create a `Counter` called `freq`. Print the full counter, then print: (i) the 3 most common words using `most_common()`, (ii) the frequency of the word "python", and (iii) the total number of words and the size of the unique vocabulary.

```python
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"
word_list = sentence.split()

freq = Counter(word_list)
print(freq)

print(freq.most_common(3))
print(freq["python"])
print(len(word_list))
print(len(freq))
```
